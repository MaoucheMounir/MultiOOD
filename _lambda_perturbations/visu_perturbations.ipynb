```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

frameworks = ["far_ood", "near_ood", "vfa"]
layer_procs = ["react", "ash"]
datasets = ["HMDB", "UCF", "EPIC"]

# Générer la liste des chemins d'images à afficher
image_paths = []
for framework in frameworks:
    if framework == "near_ood":
        for dataset in datasets:
            for layer_proc in layer_procs:
                name = "ponderation_" + framework + "_" + dataset + "_" + layer_proc + ".png"
                image_paths.append(os.path.join("..", name))
    else:
        for layer_proc in layer_procs:
            name = "ponderation_" + framework + "_" + layer_proc + ".png"
            image_paths.append(os.path.join("..", name))

# Création d'une figure avec 5 lignes, 2 colonnes
fig, axes = plt.subplots(5, 2, figsize=(12, 15))
axes = axes.flatten()  # aplatir en liste 1D

# Boucle d'affichage
for ax, path in zip(axes, image_paths):
    img = mpimg.imread(path)
    ax.imshow(img)
    ax.axis('off')
    ax.set_title(os.path.basename(path))

plt.tight_layout()
plt.savefig("fusion_images.png")
plt.show()
```
