```python
def compute_all_metrics(conf, label, pred):
    np.set_printoptions(precision=3)
    recall = 0.95
    auroc, aupr_in, aupr_out, fpr = auc_and_fpr_recall(conf, label, recall)

    accuracy = acc(pred, label)

    results = [fpr, auroc, aupr_in, aupr_out, accuracy]

    return results


# accuracy
def acc(pred, label):
    ind_pred = pred[label != -1]
    ind_label = label[label != -1]

    num_tp = np.sum(ind_pred == ind_label)
    acc = num_tp / len(ind_label)

    return acc


def auc_and_fpr_recall(conf, label, tpr_th):
    # following convention in ML we treat OOD as positive
    ood_indicator = label
    #ood_indicator = np.zeros(label)
    #ood_indicator[label == -1] = 1

    # in the postprocessor we assume ID samples will have larger
    # "conf" values than OOD samples
    # therefore here we need to negate the "conf" values
    fpr_list, tpr_list, thresholds = metrics.roc_curve(ood_indicator, -conf)
    fpr = fpr_list[np.argmax(tpr_list >= tpr_th)]

    precision_in, recall_in, thresholds_in \
        = metrics.precision_recall_curve(1 - ood_indicator, conf)

    precision_out, recall_out, thresholds_out \
        = metrics.precision_recall_curve(ood_indicator, -conf)

    auroc = metrics.auc(fpr_list, tpr_list)
    aupr_in = metrics.auc(recall_in, precision_in)
    aupr_out = metrics.auc(recall_out, precision_out)

    return auroc, aupr_in, aupr_out, fpr
```

# Max/Mean/Min SP

Pour Mean et Min:
- récupérer les fichiers 'output' (logits) et appliquer un softmax
- Prendre le mean 

```python
output_hmdb = get_result(ood_dataset=None, case="output", mode='val')
#output_ood = get_result(ood_dataset="UCF", case="output")

sm_hmdb = torch.softmax(torch.tensor(output_hmdb), dim=1)
means_sm_hmdb = torch.mean(sm_hmdb, dim=1)
min_sm_hmdb = torch.min(sm_hmdb, dim=1)
```

```python
def compute_metrics_sp(ood_dataset, pooling="max"): 
    #mode = {max, mean, min} max est équivalent à loader conf mais pour min et mean il faut loader output et faire softmax et prendre le min ou mean à la main
    assert pooling in {"max", "mean", "min"}
    
    if pooling == "max":
        conf_id = get_result(None, "conf", "val") #id est hmdb
        conf_ood = get_result(ood_dataset, "conf")

    else:
        output_id = get_result(ood_dataset=None, case="output", mode='val')
        output_ood = get_result(ood_dataset, case="output")
        
        sm_id = torch.softmax(torch.tensor(output_id), dim=1)
        sm_ood = torch.softmax(torch.tensor(output_ood), dim=1)
        if pooling == "mean":
            conf_id = torch.mean(sm_id, dim=1) 
            conf_ood = torch.mean(sm_ood, dim=1) 
            
        elif pooling == "min":
            conf_id = torch.min(sm_id, dim=1)[0]
            conf_ood = torch.min(sm_ood, dim=1)[0] 
            # print(conf_id)
            # print(conf_ood)
    
    pred_id = get_result(None, "pred", "val") # quand je fais le min, est ce que je prends comme pred le max ? jpense oui
    pred_ood = get_result(ood_dataset, "pred")
    
    conf_combine = np.hstack((conf_id, conf_ood))
    label_combine = np.hstack((np.zeros_like(conf_id), np.ones_like(conf_ood)))
    pred_combine = np.hstack((pred_id, pred_ood))
    
    
    return compute_all_metrics(conf_combine, label_combine, pred_combine) 
    # fpr, auroc, aupr_in, aupr_out, accuracy
```

```python
results = {}

for pooling in pooling_types:
    results.setdefault(pooling, {})
    
    for dataset in ood_datasets:
        results.get(pooling,-1).setdefault(dataset, {})
        fpr, auroc, aupr_in, aupr_out, accuracy = compute_metrics_sp(dataset, pooling)
        #results.get(pooling,-1).get(dataset,-1).setdefault(dataset, {})
        results.get(pooling,-1).get(dataset,-1)["auroc"] = auroc
        results.get(pooling,-1).get(dataset,-1)["fpr"] = fpr
```
