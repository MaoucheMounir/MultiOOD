```python
import pandas as pd
```

```python
ordre_method = ['msp', 'ebo', 'maxlogit', 'Mahalanobis', 'react', 'ash', 'gen', 'knn', 'vim']
ordre_dataset = ['UCF', 'EPIC', 'HAC']
ordre_layer_proc = ['none', 'ash', 'react']
```

```python
results_file = "results_postprocessings_far_ood_baseline.csv"
df_results_baseline = pd.read_csv(results_file).drop(["backbone", "layer_proc"], axis=1)
```

normaliser les valeurs et les ordres 

```python
df_results_baseline.fpr95 = df_results_baseline.fpr95.apply(lambda x:x*100)
df_results_baseline.auroc = df_results_baseline.auroc.apply(lambda x:x*100)
df_results_baseline.id_acc = df_results_baseline.id_acc.apply(lambda x:x*100)
```

afficher les colonnes comme dans le tableau du papier

```python
results = {"method": [], "UCF_fpr95": [], "UCF_auroc":[], "EPIC_fpr95": [], "EPIC_auroc": [], "HAC_fpr95":[], "HAC_auroc":[]}
metrics = ["fpr95", "auroc"]
for name, group in df_results_baseline.groupby(by="method"):
    
    results["method"].append(name)
    for i, dataset in group.iterrows():
        for metric in metrics:
            results[
                f"{dataset['dataset']}_{metric}"
                ].append(dataset[metric])
        

ordre_method = ['msp', 'ebo', 'maxlogit', 'Mahalanobis', 'react', 'ash', 'gen', 'knn', 'vim']
results = pd.DataFrame(results)
results['method'] = pd.Categorical(results['method'], categories=ordre_method, ordered=True)
results = results.sort_values(['method'])
results.to_csv("tableau7_far_ood_hmdb_baseline.csv", index=False)
results
```
