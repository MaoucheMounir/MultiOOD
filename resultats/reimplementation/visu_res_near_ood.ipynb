```python
import pandas as pd
```

```python
ordre_method = ['msp', 'ebo', 'maxlogit', 'Mahalanobis', 'react', 'ash', 'gen', 'knn', 'vim']
ordre_dataset = ['UCF', 'EPIC', 'HAC']
ordre_layer_proc = ['none', 'ash', 'react']
```

```python
results_file = "results_postprocessings_near_ood.csv"
df_results = pd.read_csv(results_file)
```

```python
df_results_baseline = df_results[df_results.backbone == 'baseline'].drop("backbone", axis=1) 
df_results_a2dnpm = df_results[df_results.backbone == 'a2d_npmix'].drop("backbone", axis=1) 
```

normaliser les valeurs et les ordres 

```python
df_results_baseline['method'] = pd.Categorical(df_results_baseline['method'], categories=ordre_method, ordered=True)
df_results_baseline['layer_proc'] = pd.Categorical(df_results_baseline['layer_proc'], categories=ordre_layer_proc, ordered=True)

# Trier d'abord par 'method', puis par 'dataset'
df_results_baseline = df_results_baseline.sort_values(['method', 'layer_proc']) #['method', 'dataset', 'layer_proc']
df_results_baseline.fpr95 = df_results_baseline.fpr95.apply(lambda x:x*100)
df_results_baseline.auroc = df_results_baseline.auroc.apply(lambda x:x*100)
df_results_baseline.id_acc = df_results_baseline.id_acc.apply(lambda x:x*100)
```

```python
df_results_a2dnpm['method'] = pd.Categorical(df_results_a2dnpm['method'], categories=ordre_method, ordered=True)
df_results_a2dnpm['layer_proc'] = pd.Categorical(df_results_a2dnpm['layer_proc'], categories=ordre_layer_proc, ordered=True)

# Trier d'abord par 'method', puis par 'dataset'
df_results_a2dnpm = df_results_a2dnpm.sort_values(['method', 'layer_proc']) #['method', 'dataset', 'layer_proc']
df_results_a2dnpm.fpr95 = df_results_a2dnpm.fpr95.apply(lambda x:x*100)
df_results_a2dnpm.auroc = df_results_a2dnpm.auroc.apply(lambda x:x*100)
df_results_a2dnpm.id_acc = df_results_a2dnpm.id_acc.apply(lambda x:x*100)
```

afficher les colonnes comme dans le tableau du papier

```python
table1_baseline = {"method": [], "HMDB_fpr95":[], "HMDB_auroc": [],
           "UCF_fpr95": [], "UCF_auroc": [], "EPIC_fpr95": [], "EPIC_auroc": []}#, "HAC_fpr95":[], "HAC_auroc":[]}
metrics = ["fpr95", "auroc"]
for name, group in df_results_baseline.groupby(by="method"):
    
    table1_baseline["method"].append(name)
    for i, dataset in group.iterrows():
        for metric in metrics:
            table1_baseline[
                f"{dataset['dataset']}_{metric}"
                ].append(dataset[metric])
        

ordre_method = ['msp', 'ebo', 'maxlogit', 'Mahalanobis', 'react', 'ash', 'gen', 'knn', 'vim']
table1_baseline = pd.DataFrame(table1_baseline)
table1_baseline['method'] = pd.Categorical(table1_baseline['method'], categories=ordre_method, ordered=True)
table1_baseline = table1_baseline.sort_values(['method'])
table1_baseline.to_csv("tableau1_near_ood_baseline.csv")
table1_baseline
```

```python
table1_a2dnpmix = {"method": [], "HMDB_fpr95":[], "HMDB_auroc": [],
           "UCF_fpr95": [], "UCF_auroc": [], "EPIC_fpr95": [], "EPIC_auroc": []}#, "HAC_fpr95":[], "HAC_auroc":[]}
metrics = ["fpr95", "auroc"]
for name, group in df_results_a2dnpm.groupby(by="method"):
    
    table1_a2dnpmix["method"].append(name)
    for i, dataset in group.iterrows():
        for metric in metrics:
            table1_a2dnpmix[
                f"{dataset['dataset']}_{metric}"
                ].append(dataset[metric])
        

ordre_method = ['msp', 'ebo', 'maxlogit', 'Mahalanobis', 'react', 'ash', 'gen', 'knn', 'vim']
table1_a2dnpmix = pd.DataFrame(table1_a2dnpmix)
table1_a2dnpmix['method'] = pd.Categorical(table1_a2dnpmix['method'], categories=ordre_method, ordered=True)
table1_a2dnpmix = table1_a2dnpmix.sort_values(['method'])
table1_a2dnpmix.to_csv("tableau1_near_ood_a2dnpmix.csv", index=False)
table1_a2dnpmix
```
