# Far OOD Video Flow

```python
df_results = pd.read_csv( "results_moda_wise_ucf_react_video_flow.csv")
df_results = normalize_columns(drop_columns(df_results, ["backbone", "method", "layer_proc", "exec_time"])
                               , ["fpr95", "auroc","id_acc"])
df_results
```

afficher les colonnes comme dans le tableau du papier

```python
results = {"modality": [], "UCF_fpr95": [], "UCF_auroc":[], "EPIC_fpr95": [], "EPIC_auroc": [], "HAC_fpr95":[], "HAC_auroc":[]}
metrics = ["fpr95", "auroc"]
for name, group in df_results.groupby(by="drop_modality"):
    results["modality"].append(name)
    for i, dataset in group.iterrows():
        for metric in metrics:
            results[
                f"{dataset['dataset']}_{metric}"
                ].append(dataset[metric])
        

#ordre_method = ['msp', 'ebo', 'maxlogit', 'Mahalanobis', 'react', 'ash', 'gen', 'knn', 'vim']
ordre_modalites = ["video", "flow"]
results = pd.DataFrame(results)
results['modality'] = pd.Categorical(results['modality'], categories=ordre_modalites, ordered=True)
results = results.sort_values(['modality'])
results.to_csv("react_modality_wise_far_ood_baseline.csv", index=False)
results
```
